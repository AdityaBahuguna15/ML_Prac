**Activation functions:**  
They apply a non linear transformation and decide whether a neuron should be activated or not.  
f ($w^T.x$) -> Activation fnct  

**Why Activation functions:**  
- O = $w^T.x$  -> Linear Transforms  
&nbsp;>O  
&emsp;&emsp;>O -> y  
&nbsp;>O  
 
- Without them, network is basically a stacked linear regression model
- With non linear transformations, network can learn better and perform more complex tasks
- After each layer we typically use an activation fnct  

**Popular activation functions:**
1) Step Function:  
$f(x) = \begin{cases} 1 & \text{if } x \geq \theta \\ 0 & \text{otherwise} \end{cases}$  
- Ouputs 1, if greater than a threshold, otherwise 0
- Isnt used much 

2) Sigmoid Function:  
$f(x) = \frac {1}{1+e^{-x}}$  
- Outputs a probability btwn 0 and 1  
- Typically used in last layer of binary classification problem 

3) TanH function:  
$f(x) = \frac {2}{1+e^{-2x}}-1$ 
- Scaled Sigmoid function
- Outputs a value btwn -1,1
- Hidden layers

4) ReLU function:  
$f(x) = \max(0,x)$
- Outputs 0 for negative values, keeps +ve values same
- Very popular in networks,
- If you dont know what to use, use ReLU for hidden layers

5) Leaky ReLU function:  
$f(x) = \begin{cases} x & \text{if } x \geq 0 \\ a.x & \text{otherwise} \end{cases}$
- Slightly improved version of ReLU
- Multiplies -ve values with a very small number, leaving +ve numbers untouched
- Tries to solve the vanishing gradient problem, where with a normal ReLU, -ve values are 0, means gradient in backpropogation is zero, leading to non-updated weights, neurons wont learn anything

6) Softmax Function:  
$S(y_i) = \frac {e^{y_i}}{ \sum {e^{y_j}}}$
- Squash the inputs to be outputs btwn 0-1 (probability)
- Good to use in last layer of a multiclass classification problem

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
# option 1 (create nn modules)
class NeuralNet(nn.Module):
    def __init__(self, input_size, hidden_size):
        super(NeuralNet, self).__init__()
        self.linear1 = nn.Linear(input_size, hidden_size) # Linear layer
        self.relu = nn.ReLU() # Activation Layer
        self.linear2 = nn.Linear(hidden_size, 1) # Linear layer
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.linear1(x)
        out = self.relu(out)
        out = self.linear2(out)
        out = self.sigmoid(out)
        return out

# option 2 (use activation functions directly in forward pass)
class NeuralNet(nn.Module):
    def __init__(self, input_size, hidden_size):
        super(NeuralNet, self).__init__()
        self.linear1 = nn.Linear(input_size, hidden_size) # Linear layer
        self.linear2 = nn.Linear(hidden_size, 1) # Linear layer

    def forward(self, x):
        out = torch.relu(self.linear1(x))
        out = torch.sigmoid(self.linear2(out))
        return out

# Different Fncts:

# nn modules:
# nn.Sigmoid
# nn.Softmax
# nn.Tanh
# nn.LeakyReLU

# pytorch api:
# F.leaky_relu()
```
